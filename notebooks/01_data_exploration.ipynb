# Data Exploration
Exploratory analysis of NOAA Storm Events, EAGLE-I outage data, and ERCOT grid load.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('dark_background')
sns.set_palette('husl')

# Load NOAA storm events
storms = pd.read_csv('../data/raw/storm_events_details.csv', low_memory=False)
print(f'Storm events: {len(storms):,} rows, {len(storms.columns)} columns')
print(f'Date range: {storms["BEGIN_DATE_TIME"].min()} to {storms["BEGIN_DATE_TIME"].max()}')
storms.head()
```

```python
# Event type distribution
fig, ax = plt.subplots(figsize=(12, 6))
storms['EVENT_TYPE'].value_counts().head(15).plot(kind='barh', ax=ax, color='#3b82f6')
ax.set_xlabel('Count')
ax.set_title('Top 15 Weather Event Types (NOAA Storm Events)')
plt.tight_layout()
plt.savefig('../paper/figures/event_type_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Geographic distribution of events
fig, ax = plt.subplots(figsize=(10, 6))
valid_geo = storms.dropna(subset=['BEGIN_LAT', 'BEGIN_LON'])
ax.scatter(valid_geo['BEGIN_LON'], valid_geo['BEGIN_LAT'], s=0.5, alpha=0.3, c='#f97316')
ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')
ax.set_title('Geographic Distribution of Storm Events')
plt.tight_layout()
plt.savefig('../paper/figures/event_geographic_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Monthly event frequency
storms['BEGIN_DATE_TIME'] = pd.to_datetime(storms['BEGIN_DATE_TIME'], format='mixed', errors='coerce')
monthly = storms.groupby(storms['BEGIN_DATE_TIME'].dt.month).size()

fig, ax = plt.subplots(figsize=(10, 5))
monthly.plot(kind='bar', ax=ax, color='#22c55e')
ax.set_xlabel('Month')
ax.set_ylabel('Event Count')
ax.set_title('Monthly Distribution of Storm Events')
ax.set_xticklabels(['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec'], rotation=45)
plt.tight_layout()
plt.savefig('../paper/figures/monthly_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```
